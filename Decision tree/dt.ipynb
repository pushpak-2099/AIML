```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
print(iris)
```

```python
print(iris['DESCR'])

#based on this four features we need to predict Iris-Setosa, versicolour, virginica
```

```python
print(iris['target'])
```

```python
#independent feature 
X=pd.DataFrame(iris['data'],columns=['sepal length','sepal width','petal length','petal width'])
X
```

```python
#dependent feature 
y=pd.DataFrame(iris['target'])
```

```python
#train test split 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
#dt model
from sklearn.tree import DecisionTreeClassifier

dtclassifier=DecisionTreeClassifier()
dtclassifier.fit(X_train,y_train)
```

```python
# tree visualisation 

from sklearn import tree
plt.figure(figsize=[15,10])
tree.plot_tree(dtclassifier,filled=True)
```

```python
# get the resut (no hyperparameter tuning)
y_pred=dtclassifier.predict(X_test)

#checking the preformance 
from sklearn.metrics import classification_report,confusion_matrix

report=classification_report(y_test,y_pred)
matrix=confusion_matrix(y_test,y_pred)

print(report)

print()
print(matrix)
```

```python
#pruning (post pruning)

dtclassifier=DecisionTreeClassifier(max_depth=2)
dtclassifier.fit(X_train,y_train)
```

```python

from sklearn import tree
plt.figure(figsize=[15,10])
tree.plot_tree(dtclassifier,filled=True)
```

```python
# this much tree is enough for the predicting the result 
y_pred=dtclassifier.predict(X_test)

#checking the preformance 
from sklearn.metrics import classification_report,confusion_matrix

report=classification_report(y_test,y_pred)
matrix=confusion_matrix(y_test,y_pred)

print(report)

print()
print(matrix)
```

Hyperparameter tuning 

```python
# criterion{“gini”, “entropy”, “log_loss”}, default=”gini”
# splitter{“best”, “random”}, default=”best”
# max_featuresint, float or {“sqrt”, “log2”}, default=None

param={
    'criterion':['gini', 'entropy','log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4],
    'max_features' : ['sqrt', 'log2']


}

param
```

```python
from sklearn.model_selection import GridSearchCV

treemodel=DecisionTreeClassifier()

grid=GridSearchCV(treemodel,param_grid=param,cv=5,scoring='accuracy')
grid.fit(X,y)
```
